```python
# Import all necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("✅ Libraries imported successfully")
```

```python
# Load the raw dataset
df = pd.read_csv('../data/raw/adult.csv')

print(f"Dataset Shape: {df.shape}")
print(f"Total Rows: {df.shape[0]:,}")
print(f"Total Columns: {df.shape[1]}")
print("\n" + "="*60)
print("✅ Dataset loaded successfully")
```

```python
# Display first few rows
print("📊 FIRST 5 ROWS:")
print("="*60)
display(df.head())

print("\n📊 LAST 5 ROWS:")
print("="*60)
display(df.tail())

print("\n📊 RANDOM SAMPLE:")
print("="*60)
display(df.sample(5))
```

```python
# Get detailed column information
print("📋 DATASET INFO:")
print("="*60)
df.info()

print("\n📋 COLUMN NAMES:")
print("="*60)
print(df.columns.tolist())
```

```python
# Identify numerical and categorical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

# Remove target from categorical if present
if 'income' in categorical_cols:
    categorical_cols.remove('income')
    target_col = 'income'

print("🔢 NUMERICAL FEATURES:")
print("="*60)
print(f"Count: {len(numerical_cols)}")
print(f"Features: {numerical_cols}")

print("\n📝 CATEGORICAL FEATURES:")
print("="*60)
print(f"Count: {len(categorical_cols)}")
print(f"Features: {categorical_cols}")

print("\n🎯 TARGET VARIABLE:")
print("="*60)
print(f"Column: {target_col}")
```

```python
# Statistical summary for numerical features
print("📊 STATISTICAL SUMMARY - NUMERICAL FEATURES:")
print("="*60)
display(df[numerical_cols].describe())

# Additional statistics
print("\n📊 Skewness:")
print("="*60)
stats_df = pd.DataFrame({
    'Skewness': df[numerical_cols].skew()
})
display(stats_df)
```

```python
# Check for missing values (including '?' as missing)
print("❓ MISSING VALUES ANALYSIS:")
print("="*60)

# Traditional missing values
missing_traditional = df.isnull().sum()

# Check for '?' as missing values in categorical columns
missing_question = pd.Series(index=df.columns, dtype=int)
for col in df.columns:
    missing_question[col] = (df[col] == "?").sum()

# Combine both
missing_df = pd.DataFrame({
    'Null Values': missing_traditional,
    '?': missing_question,
    'Total': missing_traditional + missing_question,
    'Percentage': ((missing_traditional + missing_question) / len(df) * 100).round(2)
})

if len(missing_df) > 0:
    display(missing_df)
else:
    print("✅ No missing values found!")
```

```python
# Create missing values heatmap
print("🗺️ MISSING VALUES HEATMAP:")
print("="*60)

# Create a copy and mark '?' as NaN for visualization
df_missing = df.copy()
for col in df_missing.columns:
    df_missing[col] = df_missing[col].replace('?', np.nan)

plt.figure(figsize=(14, 8))
sns.heatmap(df_missing.isnull(), cbar=True,cmap='viridis' ,yticklabels=False)
plt.title('Missing Values Heatmap (Yellow = Missing)', fontsize=16, fontweight='bold')
plt.xlabel('Features', fontsize=12)
plt.ylabel('Samples', fontsize=12)
plt.tight_layout()
plt.savefig('../notebooks/missing_values_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Heatmap saved as 'missing_values_heatmap.png'")
```

```python
# Analyze target variable distribution
print("🎯 TARGET VARIABLE DISTRIBUTION:")
print("="*60)

target_counts = df['income'].value_counts()
target_pct = df['income'].value_counts(normalize=True) * 100

target_df = pd.DataFrame({
    'Class': target_counts.index,
    'Count': target_counts.values,
    'Percentage': target_pct.values
})
display(target_df)

# Visualize target distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Count plot
axes[0].bar(target_counts.index, target_counts.values, color=['#3498db', '#e74c3c'])
axes[0].set_xlabel('Income Class', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Count', fontsize=12, fontweight='bold')
axes[0].set_title('Target Distribution (Count)', fontsize=14, fontweight='bold')
for i, v in enumerate(target_counts.values):
    axes[0].text(i, v + 500, str(v), ha='center', fontweight='bold')

# Pie chart
axes[1].pie(target_counts.values, labels=target_counts.index, autopct='%1.2f%%', 
            colors=['#3498db', '#e74c3c'], startangle=90, textprops={'fontsize': 12, 'fontweight': 'bold'})
axes[1].set_title('Target Distribution (Percentage)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('../notebooks/target_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"\n⚠️ Class Imbalance Ratio: {target_pct.values[0]/target_pct.values[1]:.2f}:1")
print("✅ Target distribution visualized and saved")
```

```python
# Distribution plots for numerical features
print("📊 NUMERICAL FEATURES DISTRIBUTION:")
print("="*60)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.ravel()

for idx, col in enumerate(numerical_cols):
    axes[idx].hist(df[col], bins=30, color='#3498db', edgecolor='black', alpha=0.7)
    axes[idx].set_xlabel(col, fontsize=11, fontweight='bold')
    axes[idx].set_ylabel('Frequency', fontsize=11, fontweight='bold')
    axes[idx].set_title(f'Distribution of {col}', fontsize=12, fontweight='bold')
    axes[idx].grid(axis='y', alpha=0.3)
    
    # Add statistics
    mean_val = df[col].mean()
    median_val = df[col].median()
    axes[idx].axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.1f}')
    axes[idx].axvline(median_val, color='green', linestyle='--', linewidth=2, label=f'Median: {median_val:.1f}')
    axes[idx].legend(fontsize=9)

plt.tight_layout()
plt.savefig('../notebooks/numerical_distributions.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Numerical distributions visualized and saved")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Filter non-zero values
capital_gain_nonzero = df[df['capital.loss'] != 0]['capital.loss']

# IQR Calculation
Q1 = capital_gain_nonzero.quantile(0.25)
Q3 = capital_gain_nonzero.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Outlier count
outliers = capital_gain_nonzero[
    (capital_gain_nonzero < lower_bound) | 
    (capital_gain_nonzero > upper_bound)
]

outlier_count = len(outliers)

# Plot
plt.figure(figsize=(8,5))
sns.boxplot(x=capital_gain_nonzero)

plt.title("Capital Gain (Non-Zero Values)")
plt.xlabel("Capital Gain")

# Annotate outlier count
plt.text(
    x=0.95, 
    y=0.9, 
    s=f"Outliers: {outlier_count}",
    transform=plt.gca().transAxes,
    horizontalalignment='right',
    bbox=dict(facecolor='white', alpha=0.6)
)

plt.show()
```

```python
# Box plots to detect outliers in numerical features
print("📦 BOX PLOTS - OUTLIER DETECTION:")
print("="*60)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.ravel()

for idx, col in enumerate(numerical_cols):
    axes[idx].boxplot(df[col], vert=True, patch_artist=True,
                     boxprops=dict(facecolor='#3498db', alpha=0.7),
                     medianprops=dict(color='red', linewidth=2),
                     whiskerprops=dict(color='black', linewidth=1.5),
                     capprops=dict(color='black', linewidth=1.5))
    axes[idx].set_ylabel(col, fontsize=11, fontweight='bold')
    axes[idx].set_title(f'Box Plot - {col}', fontsize=12, fontweight='bold')
    axes[idx].grid(axis='y', alpha=0.3)
    
    # Calculate outliers using IQR method
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = ((df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR))).sum()
    axes[idx].text(0.5, 0.95, f'Outliers: {outliers}', transform=axes[idx].transAxes,
                  fontsize=10, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig('../notebooks/outlier_boxplots.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Outlier detection completed and saved")
```

```python
# Distribution of categorical features
print("📝 CATEGORICAL FEATURES DISTRIBUTION:")
print("="*60)

# Show unique counts
cat_unique = pd.DataFrame({
    'Feature': categorical_cols,
    'Unique_Values': [df[col].nunique() for col in categorical_cols],
    'Most_Common': [df[col].mode()[0] for col in categorical_cols],
    'Most_Common_Count': [df[col].value_counts().iloc[0] for col in categorical_cols]
})
display(cat_unique)

# Visualize top categorical features (those with reasonable unique counts)
features_to_plot = [col for col in categorical_cols if df[col].nunique() <= 15]

n_features = len(features_to_plot)
n_cols = 3
n_rows = (n_features + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, n_rows * 4))
axes = axes.ravel() if n_features > 1 else [axes]

for idx, col in enumerate(features_to_plot):
    value_counts = df[col].value_counts().head(10)
    axes[idx].barh(range(len(value_counts)), value_counts.values, color='#2ecc71')
    axes[idx].set_yticks(range(len(value_counts)))
    axes[idx].set_yticklabels(value_counts.index, fontsize=9)
    axes[idx].set_xlabel('Count', fontsize=11, fontweight='bold')
    axes[idx].set_title(f'Distribution of {col}', fontsize=12, fontweight='bold')
    axes[idx].grid(axis='x', alpha=0.3)
    
    # Add value labels
    for i, v in enumerate(value_counts.values):
        axes[idx].text(v + 100, i, str(v), va='center', fontweight='bold', fontsize=9)

# Hide extra subplots
for idx in range(len(features_to_plot), len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.savefig('../notebooks/categorical_distributions.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Categorical distributions visualized and saved")
```

```python
# Correlation matrix for numerical features
print("🔗 CORRELATION MATRIX - NUMERICAL FEATURES:")
print("="*60)

correlation_matrix = df[numerical_cols].corr()
display(correlation_matrix)

# Visualize correlation matrix
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8},
            vmin=-1, vmax=1, center=0)
plt.title('Correlation Matrix - Numerical Features', fontsize=16, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()
plt.savefig('../notebooks/correlation_matrix.png', dpi=300, bbox_inches='tight')
plt.show()

# Find highly correlated pairs
print("\n🔍 HIGHLY CORRELATED FEATURE PAIRS (|correlation| > 0.7):")
print("="*60)
high_corr = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        if abs(correlation_matrix.iloc[i, j]) > 0.7:
            high_corr.append({
                'Feature_1': correlation_matrix.columns[i],
                'Feature_2': correlation_matrix.columns[j],
                'Correlation': correlation_matrix.iloc[i, j]
            })

if high_corr:
    display(pd.DataFrame(high_corr))
else:
    print("✅ No highly correlated pairs found (threshold > 0.7)")
```

```python
# Analyze relationship between numerical features and target
print("🎯 NUMERICAL FEATURES vs TARGET:")
print("="*60)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.ravel()

for idx, col in enumerate(numerical_cols):
    for income_class in df['income'].unique():
        subset = df[df['income'] == income_class][col]
        axes[idx].hist(subset, bins=20, alpha=0.6, label=income_class, edgecolor='black')
    
    axes[idx].set_xlabel(col, fontsize=11, fontweight='bold')
    axes[idx].set_ylabel('Frequency', fontsize=11, fontweight='bold')
    axes[idx].set_title(f'{col} Distribution by Income', fontsize=12, fontweight='bold')
    axes[idx].legend()
    axes[idx].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('../notebooks/target_vs_numerical.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Target vs numerical features visualized")
```

```python
# Analyze relationship between categorical features and target
print("🎯 CATEGORICAL FEATURES vs TARGET:")
print("="*60)

features_to_plot = [col for col in categorical_cols if df[col].nunique() <= 10]

n_features = len(features_to_plot)
n_cols = 2
n_rows = (n_features + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, n_rows * 5))
axes = axes.ravel() if n_features > 1 else [axes]

for idx, col in enumerate(features_to_plot):
    # Create crosstab
    ct = pd.crosstab(df[col], df['income'], normalize='index') * 100
    ct.plot(kind='bar', ax=axes[idx], color=['#3498db', '#e74c3c'], width=0.8)
    axes[idx].set_xlabel(col, fontsize=11, fontweight='bold')
    axes[idx].set_ylabel('Percentage (%)', fontsize=11, fontweight='bold')
    axes[idx].set_title(f'{col} vs Income', fontsize=12, fontweight='bold')
    axes[idx].legend(title='Income', fontsize=9)
    axes[idx].grid(axis='y', alpha=0.3)
    axes[idx].tick_params(axis='x', rotation=45)

# Hide extra subplots
for idx in range(len(features_to_plot), len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.savefig('../notebooks/target_vs_categorical.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Target vs categorical features visualized")
```

```python
# Detailed outlier analysis using IQR method
print("📊 OUTLIER STATISTICS (IQR Method):")
print("="*60)

outlier_stats = []

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
    
    outlier_stats.append({
        'Feature': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': len(outliers),
        'Outlier_Percentage': (len(outliers) / len(df) * 100)
    })

outlier_df = pd.DataFrame(outlier_stats)
display(outlier_df)

print("\n✅ Outlier analysis completed")
```

```python
# Analyze skewness of numerical features
print("📐 SKEWNESS ANALYSIS:")
print("="*60)

skewness_df = pd.DataFrame({
    'Feature': numerical_cols,
    'Skewness': df[numerical_cols].skew().values,
    'Interpretation': ['Highly Left Skewed' if x < -1 else 'Left Skewed' if x < -0.5 else 
                      'Symmetric' if abs(x) <= 0.5 else 'Right Skewed' if x < 1 else 'Highly Right Skewed' 
                      for x in df[numerical_cols].skew().values]
})
display(skewness_df)

# Visualize skewness
plt.figure(figsize=(12, 6))
colors = ['red' if abs(x) > 1 else 'orange' if abs(x) > 0.5 else 'green' 
          for x in skewness_df['Skewness']]
plt.barh(skewness_df['Feature'], skewness_df['Skewness'], color=colors, alpha=0.7)
plt.xlabel('Skewness Value', fontsize=12, fontweight='bold')
plt.ylabel('Features', fontsize=12, fontweight='bold')
plt.title('Skewness of Numerical Features', fontsize=14, fontweight='bold')
plt.axvline(x=0, color='black', linestyle='-', linewidth=1)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.savefig('../notebooks/skewness_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Skewness analysis completed")
```
